```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification

X, y = make_classification(
    n_samples=200,
    n_features=2,
    n_informative=2,
    n_redundant=0,
    n_clusters_per_class=1,
    class_sep=3.0,
    random_state=3
)

plt.scatter(X[:, 0], X[:, 1], c=y, cmap="bwr")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Linearly Separable Binary Classification Dataset")
plt.show()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.2,random_state=2)
```

```python
import pandas as pd
df = pd.DataFrame(X,y)
df
```

```python
epochs = 1000
lr = 0.01
```

```python
def perceptron1(X,y):
  X = np.insert(X,0,1,axis=1)
  weights = np.ones(X.shape[1])
  lr = 0.1

  for i in range(epochs):
    idx = np.random.randint(0,X.shape[0])
    prod = np.dot(X[idx],weights)
    y_hat = 1 if prod > 0 else 0
    weights = weights + lr*(y[idx]-y_hat)*X[idx]
  return weights[0],weights[1:]
```

```python
intercept,coef = perceptron1(X_train,y_train)
intercept,coef
```

```python
m = -(coef[0]/coef[1])
b = -(intercept/coef[1])
x_input = np.linspace(-6,6,100)
y_input = m*x_input + b
```

```python
plt.figure(figsize=(10, 6))
plt.plot(x_input, y_input, color='red', label='Decision Boundary')
plt.scatter(X[:, 0], X[:, 1], c=y, cmap="bwr")

# Limit the view to focus on the actual range of your dataset
plt.xlim(X[:, 0].min() - 1, X[:, 0].max() + 1)
plt.ylim(X[:, 1].min() - 1, X[:, 1].max() + 1)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Linearly Separable Binary Classification with Perceptron Boundary")
plt.legend()
plt.show()
```
