```python
class GDRegressor:
  def __init__(self,learning_rate,epochs) -> None:
    self.m = 100
    self.b = -120
    self.learning_rate = learning_rate
    self.epochs = epochs

  def fit(self,X,y):
    for i in range(self.epochs):
      loss_slope_b = -2 * np.sum(y - self.m*X.ravel()-self.b)
      loss_slope_m = -2 * np.sum((y - self.m*X.ravel()-self.b)*X.ravel())
      self.b -= self.learning_rate*loss_slope_b
      self.m -= self.learning_rate*loss_slope_m
    return (self.m,self.b)

  def predict(self,X):
    return self.m*X + self.b
```

```python
# now will try to match these coefs and intercept value using the regression class made from scratch
```

```python
class GDRegressor:
  def __init__(self,learning_rate=0.01,epochs=100):
    self.learning_rate = learning_rate
    self.epochs = epochs
    self.coefs = None
    self.intercept = None

  def fit(self,X,y):
    #initializing intercept and coefs
    self.intercept = 0
    self.coefs = np.ones(X.shape[1])

    for i in range(self.epochs):
      y_hat = np.dot(X,self.coefs) + self.intercept

      intercept_der = -2*np.mean(y-y_hat)
      coefs_der = -2 * np.dot((y-y_hat),X)/X.shape[0]

      self.intercept -= self.learning_rate*intercept_der
      self.coefs -= self.learning_rate*coefs_der

  def predict(self,X):
    return np.dot(X,self.coefs) + self.intercept
```

```python
gdr = GDRegressor(learning_rate=0.5,epochs = 1000)
```

```python
gdr.fit(X_train,y_train)
```

```python
gdr.coefs
```

```python
gdr.intercept
```

```python
y_pred = gdr.predict(X_test)
```

```python
r2_score(y_test,y_pred)
```
